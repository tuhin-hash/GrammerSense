# Grammar Error Detection & Correction (GEC / GED) Exploration

This interactive notebook demonstrates the 6 core stages of the Grammar Error Correction pipeline:
1. **Data Inspection**: Loading source-target sentence pairs.
2. **Tokenization & Offsets**: Preserving character spans in source sentences.
3. **Dynamic Programming Alignment**: Extracting edit operations (insert, delete, replace).
4. **Error Classification**: Distinguishing directly identified rules vs inferred classifications.
5. **Candidate Generation & Selection**: Beam search generation and confidence ranking.
6. **Evaluation**: Computing sentence and corpus GLEU, BLEU, and F0.5 scores.

```python
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from backend.app.preprocessing.text_cleaner import TextCleaner
from backend.app.preprocessing.span_aligner import SpanAligner
from backend.app.services.detection_service import DetectionService
from backend.app.correction.candidate_selector import CandidateSelector
from backend.app.evaluation.metrics import GECMetrics

print("All modules loaded successfully.")
```

## 1. Span Localization & Edit Extraction

```python
source_sentence = "She go to college every day."
target_sentence = "She goes to college every day."

edits = SpanAligner.align(source_sentence, target_sentence)
for edit in edits:
    print(f"Operation: {edit.operation} | Original: '{edit.source_text}' -> Replacement: '{edit.target_text}'")
    print(f"Character span: [{edit.source_start_char}:{edit.source_end_char}]")
```

## 2. Error Classification (Directly Identified vs Inferred)

```python
detected_errors = DetectionService.detect_and_classify(source_sentence, target_sentence, edits)
for err in detected_errors:
    print(f"Category: {err.category}")
    print(f"Tier: [{err.verification_tier}]")
    print(f"Explanation: {err.explanation}")
```

## 3. Evaluation Metrics (Google GLEU & F0.5)

```python
gleu = GECMetrics.sentence_gleu(source_sentence, target_sentence, target_sentence)
print(f"Sentence GLEU score: {gleu * 100:.2f}%")
```
